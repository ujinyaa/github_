```python
# 1. 성별 라벨링
df['성별'] = df['성별'].replace({'남성': 0, '여성': 1})
# 2. 국적, 대륙 결측치 → '미상'
df[['국적', '대륙']] = df[['국적', '대륙']].fillna('미상')

# 3. 체류자격 컬럼 필터링
stay_cols = [col for col in df.columns if any(prefix in col for prefix in ['A', 'B', 'C', 'D','E','F','G','T', '기타'])]

# 4. 체류자격: NaN → 0, float → int
df[stay_cols] = df[stay_cols].apply(pd.to_numeric, errors='coerce').fillna(0).astype(int)

#대륙 리스트 추출 후 대륙별 데이터 프레임을 저장
continent_names = df['대륙'].unique()
Continent = [df[df['대륙'] == name] for name in continent_names]

# 5. 기타 국적/대륙 제거한 버전도 생성 (선택)
df_no_others = df[df['국적'] != '기타(Others)']
df_no_other_continent = df[df['대륙'] != '기타']

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_palette("pastel")

# 대륙 기준 시각화
grouped = df.groupby(df['대륙'])[stay_cols].sum()
plt.figure(figsize=(12, 6))
grouped.plot(kind='bar', stacked=True)
plt.title('대륙별 체류자격 총합 분포')
plt.xlabel('대륙')
plt.ylabel('체류 인원 수')
plt.xticks(rotation=45)
plt.tight_layout()
plt.legend(title='체류자격', bbox_to_anchor=(1.15, 1))
plt.show()
```

```python
# 2. 반복문으로 대륙별 국가 상위 N개 체류자격 plot
for continent_df, continent_name in zip(Continent, continent_names):
    # 국적별 체류자격 총합 계산
    country_total = continent_df.groupby('국적')[stay_cols].sum()
    country_total_sum = country_total.sum(axis=1).sort_values(ascending=False)
    top_countries = country_total_sum.head(6).index  # 상위 5개 국적만
    
    # 상위 국가만 필터링 후 시각화
    top_df = country_total.loc[top_countries]

    top_df.T.plot(kind='bar', figsize=(14, 6), stacked=True)
    plt.title(f"{continent_name} 대륙 - 체류자격별 상위 국가 분포")
    plt.xlabel('체류자격')
    plt.ylabel('인원 수')
    plt.xticks(rotation=45)
    plt.legend(title='국적', bbox_to_anchor=(1.02, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```

상위 5개 국적별 체류자격을 대륙별로 비교한 PLOT을 그린다면?
일단 대륙별 상위 3개의 국적별 가장 높은 체류 조건을 각 테이블로 출력하고, 그걸 기준으로 역으로 해당 데이터에 해당하는 국적을 대륙별로 출력하고 싶어


```python
import re
# F1~F6 계열 컬럼 추출
Multi_Cultural = [col for col in df.columns if re.match(r'^F\d+', col)]

# 대륙별로 F계열 체류자격 합계
Continent_Group_F = df.groupby('대륙')[Multi_Cultural].sum()

# 시각화
plt.figure(figsize=(10, 6))
Continent_Group_F.plot(kind='bar', stacked=True, colormap='Pastel1')
plt.title('다문화가정에 해당하는 체류조건에 따른 그래프')
plt.xlabel('대륙')
plt.ylabel('체류 인원 수')
plt.xticks(rotation=45)
plt.tight_layout()
plt.legend(title='F_체류자격', bbox_to_anchor=(1.15, 1))
plt.show()
```
