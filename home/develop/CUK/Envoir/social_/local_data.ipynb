```python

# 기본 전처리: 결측치 제거
criminal_df = criminal.dropna(subset=['범죄대분류', '범죄중분류'])

# '범죄대분류', '범죄중분류'를 원핫 인코딩
encoded_major = pd.get_dummies(criminal_df['범죄대분류'], prefix='대분류')
encoded_minor = pd.get_dummies(criminal_df['범죄중분류'], prefix='중분류')

# 원본 데이터프레임에 인코딩된 결과 합치기
criminal= pd.concat([criminal_df, encoded_major, encoded_minor], axis=1)

# '서울'로 시작하는 열만 선택
```

```python
criminal_df= pd.concat([criminal_df, encoded_major, encoded_minor], axis=1)

# '서울'로 시작하는 열만 선택
seoul_columns = [col for col in criminal_df.columns if str(col).startswith('서울')]
# 구 이름 정리 ("서울" 제거)
local_names = [col.replace('서울', '') for col in seoul_columns]
# 필요한 열만 선택
df_selected = criminal_df[['범죄대분류', '범죄중분류'] + seoul_columns]

# 서울시 전체 합계 열 추가
df_selected['서울시합계'] = df_selected[seoul_columns].sum(axis=1)

# '범죄대분류'로 그룹화하여 총합 계산
total_by_major = df_selected.groupby('범죄대분류')['서울시합계'].sum()

# '범죄중분류'로 그룹화하여 총합 계산
total_by_minor = df_selected.groupby('범죄중분류')['서울시합계'].sum()
```

```python
# 범죄 대분류별 총합 시각화
plt.figure(figsize=(12, 6))
total_by_major.sort_values(ascending=False).plot(kind='bar', color='orange')
plt.title('범죄 대분류별 서울시 총합')
plt.ylabel('건수')
plt.xlabel('범죄 대분류')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 범죄 중분류별 총합 시각화
plt.figure(figsize=(14, 7))
total_by_minor.sort_values(ascending=False).plot(kind='bar', color='orange')
plt.title('범죄 중분류별 서울시 총합')
plt.ylabel('건수')
plt.xlabel('범죄 중분류')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# 결과 데이터프레임 출력 (범죄 대분류별 정렬)
df_major_sorted = df_selected.sort_values(by='서울시합계', ascending=False)

# 결과 데이터프레임 출력 (범죄 중분류별 정렬)
df_minor_sorted = df_selected.sort_values(by='서울시합계', ascending=False)
print(df_major_sorted)
print(df_minor_sorted)
```

저소득_한부모가족 1)한부모가족지원법 수급자_재가보호, 시설보호, 미보호 2) 국민기초생활보장법 수급자

```python

# 2) 6행~30행(인덱스 5~29)만 사용, 결측치 0으로 채우기
lowincome_df = lowincome.iloc[5:30].copy().fillna(0)

# 3) 숫자형 변환할 컬럼 리스트
cols_num = ['재가보호','시설보호','미보호','재가보호.1','시설보호.1','미보호.1']

# 4) 쉼표 제거 → 숫자 변환 → NaN→0 → int
for c in cols_num:
    lowincome_df[c] = (
        lowincome_df[c]
        .astype(str)
        .str.replace(',', '', regex=True)
        .str.strip()
        .pipe(pd.to_numeric, errors='coerce')
        .fillna(0)
        .astype(int)
    )
```

```python
single_parent_household_count = lowincome_df[['재가보호','시설보호','미보호']].sum()
single_parent_member_count    = lowincome_df[['재가보호.1','시설보호.1','미보호.1']].sum()

# 6) “국민기초생활보장법 수급자” 지역별 정렬 (가구수: 소계.4, 가구원수: 소계.5)
#    먼저 숫자형으로 변환
lowincome_df['소계.4'] = (
    lowincome_df['소계.4'].astype(str).str.replace(',', '', regex=True)
    .pipe(pd.to_numeric, errors='coerce').fillna(0).astype(int)
)
lowincome_df['소계.5'] = (
    lowincome_df['소계.5'].astype(str).str.replace(',', '', regex=True)
    .pipe(pd.to_numeric, errors='coerce').fillna(0).astype(int)
)
basic_living_household_count = (
    lowincome_df.set_index('자치구(1)')['소계.4'].sort_values(ascending=False)
)
basic_living_member_count    = (
    lowincome_df.set_index('자치구(1)')['소계.5'].sort_values(ascending=False)
)
```

```python
sp_house = single_parent_household_count.sort_values(ascending=False)
plt.figure(figsize=(6,4))
sp_house.plot(kind='bar')
plt.title('한부모가족지원법 수급자 _ 가구수별')
plt.xlabel('분류')
plt.ylabel('가구수')
plt.tight_layout()
plt.show()
```

```python
sp_member = single_parent_member_count.sort_values(ascending=False)
plt.figure(figsize=(6,4))
sp_member.plot(kind='bar')
plt.title('한부모가족지원법 수급자 _ 가구원수별')
plt.xlabel('분류')
plt.ylabel('가구원수')
plt.tight_layout()
plt.show()
```
